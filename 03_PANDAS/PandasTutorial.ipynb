# Introduction to Pandas
Pandas(stands for Python Data Analysis) is a powerful, open-source Python library used for data manipulation, analysis, and cleaning.

* Think of it as Microsoft Excel or Google Sheets, but built entirely inside Python code.

### -> Revolves around two primary Data structures:
* Series (1D)
* DataFrame (2D)

### -> What is Pandas Used for?
* Reading and writing data from various file formats like CSV, Excel and SQL databases.
* Cleaning and preparing data (handling missing values, filtering, removing duplicates).
* Merging, joining, and reshaping datasets.
* Performing statistical analysis and descriptive statistics.
* Visualizing data quickly.

```python
import pandas as pd
```

1. Pandas Series

          A Pandas Series is a one-dimensional labeled array capable of holding data of any type. It can be created from Python lists, NumPy arrays, dictionaries, scalar values or data loaded from external sources such as CSV files, Excel files and databases. The labels associated with a Series are called its index.

```python
s = pd.Series([1,2,3,4,5])
```

```python
s
```

2. Pandas DataFrame

        Pandas DataFrame is a two-dimensional data structure with labeled axes (rows and columns). It is created by loading the datasets from existing storage which can be a SQL database, a CSV file or an Excel file. It can be created from lists, dictionaries, a list of dictionaries etc.

```python
df = pd.DataFrame({'Name': ['Mohammed Affan', 'Mohammed Faizaan Ali Khan', 'Shaik Sahil'],
                   'Age': [19, 20, 22],
                   'City': ['Bodhan', 'Auto Nagar', 'Nizam Colony']})
```

#### Merging and Concatination

1. Concatination

```python
df1 = pd.DataFrame({"Names": ["Mohammed Affan", "Shaik Sahil", "Mohammed Faizaan Ali Khan"], 'Marks': [32, 33, 35]})

df1
```

```python
df2 = pd.DataFrame({"Names": ["Mohammed Abdul Adeeb", "Shaik Cezan", "syed waseem"], 'Marks': [38, 35, 34]})
df2
```

```python
pd.concat([df1,df2])
```

2. Merging

```python
df1 = pd.DataFrame({"Names": ["Mohammed Affan", "Shaik Sahil", "Mohammed Faizaan Ali Khan"], 'Marks': [32, 33, 35]})

df1
```

```python
df2 = pd.DataFrame({"Names": ["Mohammed Affan", "Shaik Sahil", "Mohammed Faizaan Ali Khan"], 'RollNo': [504, 507, 505]})

df2
```

```python
pd.merge(df1,df2,on="Names")
```
