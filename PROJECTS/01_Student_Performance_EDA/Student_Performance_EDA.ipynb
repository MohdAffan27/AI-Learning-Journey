# Steps to clean Data


---



### Step 1 — Create a working copy

```python
df_clean = df.copy()
```

### Step 2 — Remove exact duplicate rows
Now use Pandas' drop_duplicates() method on your working copy.

```python
df_clean = df_clean.drop_duplicates()
```

### Step 3 — Verify your results
Now check the number of rows before and after cleaning, and confirm that no exact duplicates remain.

```python
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("Remaining duplicate rows in copied dataset:", df_clean.duplicated().sum())
```

### Step 4 — Investigate missing values

#### 4.1 Count missing values after duplicate removal
First, we need to check the missing values again because we've removed three rows. We shouldn't assume the counts are still the same.

```python
df_clean.isnull().sum()
```

```python
df_clean.isnull().sum().sum()
```

```python
missing_percentage = (
    df_clean.isnull().sum().sum()
    / df_clean.size
) * 100

print(f"Missing data percentage: {missing_percentage:.2f}%")
```
