# Introduction to Matplotlib
  Matplotlib is a Python library for creating static, interactive and animated visualizations from data. It provides flexible and customizable plotting functions that help in understanding data patterns, trends and relationships effectively.




---





#####Example:
Let's create a simple line plot using Matplotlib, showcasing the ease with which you can visualize data.






```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import lineStyles
import plotly.express as px

x = np.array([1,2,3])
y = np.array([4,5,6])

plt.plot(x,y)
plt.grid()
plt.show()
```

### -> Matplotlib Pyplot
Pyplot is a module within Matplotlib that provides a MATLAB-like interface for making plots. It simplifies the process of adding plot elements such as lines, images and text to the axes of the current figure.

#####Steps to Use Pyplot:
        * Import Matplotlib: Start by importing matplotlib.pyplot as plt.
        * Create Data: Prepare your data in the form of lists or arrays.
        * Plot Data: Use plt.plot() to create the plot.
        * Customize Plot: Add titles, labels and other elements using methods like plt.title(), plt.xlabel() and plt.ylabel().
        * Display Plot: Use plt.show() to display the plot.

1. Line plot / Line Chart
        

*   Line chart is one of the basic plots and can be created using plot() function.
*   It is used to represent a relationship between two data X and Y on a different axis.
*   It is commonly used to visualize trends, changes, or patterns in data over a sequence, such as days, months, or years. Line charts make it easy to compare values and observe how data changes over time.

```python
data = {
    "salaries": [45000, 46639, 48000, 51717, 52000, 52296, 54144, 59628, 60000, 61049, 63024, 71000, 80741, 85000, 86905, 89348, 92000, 93265, 93540, 93598]
}

```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.shape
```

```python
#line plot
plt.plot(df['salaries'], color= "red", marker = "o", linestyle = ":", linewidth= 2)
plt.grid()
plt.show()
```

When you pass a single column or a 1D array to plt.plot() without providing an explicit x-variable, Matplotlib automatically defaults to using the index (0, 1, 2, etc., or your custom index labels) for the horizontal x-axis.

2. Histogram

* Histogram shows the distribution of data by grouping values into bins.
* The hist() function is used to create it, with X-axis showing bins and Y-axis showing frequencies.

Multiple Plots on Same Axis

```python
x = np.array([1, 2, 3, 4])
y = x * 2

x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]

plt.plot(x, y, label='y = 2x')
plt.plot(x1, y1, '-.', label='Second Line')

plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title("Multiple Plots on Same Axis")
plt.legend()
plt.show()
```

Fill Area Between Two Lines

```python
x = np.array([1, 2, 3, 4])
y = x * 2
y1 = [3, 5, 7, 9]

plt.plot(x, y, label='y = 2x')
plt.plot(x, y1, '-.', label='y1')
plt.fill_between(x, y, y1, color='yellow', alpha=0.4)

plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title("Filled Area Between Two Lines")
plt.legend()
plt.show()
```

Plotting Trigonometric Functions

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label="sin(x)")
plt.plot(x, y_cos, label="cos(x)")
plt.xlabel("X values")
plt.ylabel("Function value")
plt.title("Trigonometric Functions")
plt.legend()
plt.show()
```

### Type 3. *Multivariate Analysis*

visualize three or more columns of data at the same time on a single graph.

-> 3 numerical column

```python
df.head()
```

```python
df["Experience"] = [0.5, 0.8, 1.2, 1.5, 2.0, 2.7, 3.1, 3.4, 4.0, 5.1, 5.5, 6.0, 6.8, 7.2, 8.5, 9.3, 10.0, 11.0, 12.5, 14.2]
```

```python
df.head()
```

-> scatter plot / bubble plot

```python
plt.scatter(sort_Age["Age"], df["salaries"], s=df["Experience"]*50,color="skyblue", edgecolors="darkblue", alpha=0.75)
plt.title("Experience vs salary vs age")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

-> 2 numerical columns and 1 categorical column

```python
plt.scatter(df["Age"], df["salaries"], c=df["dept"].map({"HR":"green","IT":"red","Finance":"blue"}), label=("HR", "IT","Finance" ))
plt.title("Experience vs salary vs age")
plt.xlabel("Age")
plt.ylabel("Salary")
# plt.legend()
plt.show()
```

```python
color = {"HR":"green","IT":"red","Finance":"blue"}

for dept, color in color.items():
  df_dept = df[df["dept"]==dept]
  plt.scatter(df_dept["Age"], df_dept["salaries"], c=color, label=dept)
plt.legend()
plt.title("Experience vs salary vs age")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
